```python
!python3 -m pip install pandas numpy
import pandas as pd
import numpy as np

# load data 
project1_df = pd.read_csv('data/bios.csv')

#BMI 
project1_df['bmi'] = project1_df['weight_kg']/((project1_df['height_cm']/100))**2
#countrywise average
countries = ['United States','France','Germany','China']
filter_df = project1_df[project1_df['NOC'].isin(countries)]

country_profile = filter_df.groupby('NOC')[['height_cm', 'weight_kg', 'bmi']].mean()
print(" average stat by country")
print(country_profile)
highest_bmi = project1_df.sort_values(by='bmi',ascending=False).head(10)
print("------------ the highest BMI Atheletes------------")
print(highest_bmi[['NOC','name','bmi']])

```
