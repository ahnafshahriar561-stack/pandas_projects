finding 2000 after olypians. 21 century olympians

```python
# 1 load data
atheletes_df = pd.read_csv('data/bios.csv')
# 2 to numeric the str of born date+ filter the year
atheletes_df['birth_year'] = pd.to_numeric(atheletes_df['born_date'].str[:4], errors='coerce')
# 3 filter for 2000+
young_blood = atheletes_df[atheletes_df['birth_year'] >=2000]
# 4 save a csv
young_blood[['name','NOC','born_date']].to_csv("21centuryolypians.csv")
```
