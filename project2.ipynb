```python
!python3 -m pip install pandas numpy
import pandas as pd
import numpy as np

project2_df = pd.read_csv('data/bios.csv')
project2_df
```

```python
##top 10 birth city 
top_cities = project2_df['born_city'].value_counts().head(10)
print("--------- top 10 birth city of olympic athletes--------")
print(top_cities)
```

```python
##france ar moddhe sobcheye beshi kotha theke ase....
french_atheletes = project2_df[project2_df['NOC'] == 'France']
top_french_cities = french_atheletes['born_city'].value_counts().head(5)
print("---------top 5 city of france---------")
print(top_french_cities)
```

```python
####proposion of atheletes  born in countri's top cities
top_city_count = top_french_cities.iloc[0]
total_french = len(french_atheletes)
percentage = (top_city_count/total_french)*100
print(f"\n {percentage} % of french olympians were born in paris")
```
