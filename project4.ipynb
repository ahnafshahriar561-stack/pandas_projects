```python
import pandas as pd

# 1. Load the athletes dataset
athletes_df = pd.read_csv('data/bios.csv')

# 2. Extract 4-digit birth and death years safely using string slicing and coercion
athletes_df['birth_year'] = pd.to_numeric(athletes_df['born_date'].str[:4], errors='coerce')
athletes_df['death_year'] = pd.to_numeric(athletes_df['died_date'].str[:4], errors='coerce')
# 3 calculate lifespan
athletes_df['lifespan'] = athletes_df['death_year']-athletes_df['birth_year']
# 4 filter lifespan more than 100
centenarians = athletes_df[athletes_df['lifespan'] >= 100]
# 5 save the file to a new csv 
centenarians[['name','NOC','birth_year','death_year', 'lifespan']].to_csv("centenarians_atheletes.csv",index = False)
```
