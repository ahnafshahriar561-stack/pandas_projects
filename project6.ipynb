```python
!python3 -m pip install pandas numpy
import pandas as pd
import numpy as np
```

```python
#load data
atheletes_df = pd.read_csv('data/bios.csv')
# 2 calculate bmi
atheletes_df['bmi'] = atheletes_df['weight_kg']/((atheletes_df['height_cm']/100)**2)
# overweight der ber kora
overweight_df = atheletes_df[atheletes_df['bmi'] > 25]
# overweight ke export kora
overweight_df.to_csv('overweight_atheletes.csv')
```
